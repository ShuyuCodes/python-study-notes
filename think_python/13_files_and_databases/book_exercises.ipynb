# Files and Databases


Most of the programs we have seen so far are **ephemeral** in the sense that they run for a short time and produce output, but when they end, their data disappears.
Each time you run an ephemeral program, it starts with a clean slate.

Other programs are **persistent**: they run for a long time (or all the time); they keep at least some of their data in long-term storage; and if they shut down and restart, they pick up where they left off.

A simple way for programs to maintain their data is by reading and writing text files.
A more versatile alternative is to store data in a database.
Databases are specialized files that can be read and written more efficiently than text files, and they provide additional capabilities.

In this chapter, we'll write programs that read and write text files and databases, and as an exercise you'll write a program that searches a collection of photos for duplicates.
But before you can work with a file, you have to find it, so we'll start with file names, paths, and directories.

## 13.2. f-strings

```python
"""
## f-strings

One way for programs to store data is to write it to a text file.
For example, suppose you are a camel spotter, and you want to record the number of camels you have seen during a period of observation.
And suppose that in one and a half years, you have spotted `23` camels.
The data in your camel-spotting book might look like this."""
```

```python
num_years = 1.5
num_camels = 23
```

```python
"""
To write this data to a file, you can use the `write` method, which we saw in [Chapter 8](section_writing_files).
The argument of `write` has to be a string, so if we want to put other values in a file, we have to convert them to strings.
The easiest way to do that is with the built-in function `str`.

Here's what that looks like:
"""

writer = open('camel-spotting-book.txt', 'w')
writer.write(str(num_years))
writer.write(str(num_camels))
writer.close()
```

### 13.4. Shelve

A **database** is a file that is organized for storing data.
Some databases are organized like a table with rows and columns of information.
Others are organized like a dictionary that maps from keys to values; they are sometimes called **key-value stores**.

The `shelve` module provides functions for creating and updating a key-value store called a "shelf".
As an example, we'll create a shelf to contain captions for the figures in the `photos` directory.
We'll use the `config` dictionary to get the name of the directory where we should put the shelf.

## 13.5. Storing data structures

In the previous example, the keys and values in the shelf are strings.
But we can also use a shelf to contain data structures like lists and dictionaries.

As an example, let's revisit the anagram example from an exercise in [Chapter 11](section_exercise11).
Recall that we made a dictionary that maps from a sorted string of letters to the
list of words that can be spelled with those letters.
For example, the key `'opst'` maps to the list `['opts', 'post', 'pots', 'spot', 'stop', 'tops']`.

We'll use the following function to sort the letters in a word.

```python
def sort_word(word):
    return ''.join(sorted(word))
```

```python
# And here’s an example.

word = 'pots'
key = sort_word(word)
key
```

```python
# Now let’s open a shelf called anagram_map. The argument 'n' means we should always create a new, empty shelf, even if one already exists.

db = shelve.open('anagram_map', 'n')
```

```python
# Now we can add an item to the shelf like this.

db[key] = [word]
db[key]
```

```python
"""
In this item, the key is a string and the value is a list of strings.

Now suppose we find another word that contains the same letters, like tops"""

word = 'tops'
key = sort_word(word)
key
```

```python
# The key is the same as in the previous example, so we want to append a second word to the same list of strings. Here’s how we would do it if db were a dictionary.

db[key].append(word)          # INCORRECT
```

```python
# But if we run that and then look up the key in the shelf, it looks like it has not been updated.

db[key]
```

```python
# Here’s the problem: when we look up the key, we get a list of strings, but if we modify the list of strings, it does not affect the shelf. If we want to update the shelf, we have to read the old value, update it, and then write the new value back to the shelf.

anagram_list = db[key]
anagram_list.append(word)
db[key] = anagram_list
```

```python
# Now the value in the shelf is updated.

db[key]
```

## 13.6. Checking for equivalent files

Now let’s get back to the goal of this chapter: searching for different files that contain the same data. One way to check is to read the contents of both files and compare.

If the files contain images, we have to open them with mode 'rb', where 'r' means we want to read the contents and 'b' indicates binary mode. In binary mode, the contents are not interpreted as text – they are treated as a sequence of bytes.

Here’s an example that opens and reads an image file.



```python
path1 = 'photos/jan-2023/photo1.jpg'
data1 = open(path1, 'rb').read()
type(data1)
```

```python
"""
The result from read is a bytes object – as the name suggests, it contains a sequence of bytes.

In general the contents of an image file and not human-readable. But if we read the contents from a second file, we can use the == operator to compare.

"""

path2 = 'photos/jan-2023/photo2.jpg'
data2 = open(path2, 'rb').read()
data1 == data2
```

```python
# These two files are not equivalent.

# Let’s encapsulate what we have so far in a function.

def same_contents(path1, path2):
    data1 = open(path1, 'rb').read()
    data2 = open(path2, 'rb').read()
    return data1 == data2
```

```python
"""
If we have only two files, this function is a good option. But suppose we have a large number of files and we want to know whether any two of them contain the same data. It would be inefficient to compare every pair of files.

An alternative is to use a hash function, which takes the contents of a file and computes a digest, which is usually a large integer. If two files contain the same data, they will have the same digest. If two files differ, they will almost always have different digests.

The hashlib module provides several hash functions – the one we’ll use is called md5. We’ll start by using hashlib.md5 to create a HASH object."""

import hashlib

md5_hash = hashlib.md5()
type(md5_hash)
```

```python
# The HASH object provides an update method that takes the contents of the file as an argument.

md5_hash.update(data1)
```

```python
# Now we can use hexdigest to get the digest as a string of hexadecimal digits that represent an integer in base 16.

digest = md5_hash.hexdigest()
digest
```

```python
# The following function encapsulates these steps.

def md5_digest(filename):
    data = open(filename, 'rb').read()
    md5_hash = hashlib.md5()
    md5_hash.update(data)
    digest = md5_hash.hexdigest()
    return digest
```

```python
# If we hash the contents of a different file, we can confirm that we get a different digest.

filename2 = 'photos/feb-2023/photo2.jpg'
md5_digest(filename2)
```

## 13.7. Walking directories

The following function takes as an argument the directory we want to search. It uses listdir to loop through the contents of the directory. When it finds a file, it prints its complete path. When it finds a directory, it calls itself recursively to search the subdirectory.

```python
def walk(dirname):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        if os.path.isfile(path):
            print(path)
        elif os.path.isdir(path):
            walk(path)
```

```python
# We can use it like this:

walk('photos')  # The order of the results depends on details of the operating system.
```

```python
# Here is a more general version of `walk` that takes as a second parameter a function object.
# Instead of printing the path of the files it discovers, it calls this function and passes the path as a parameter.

# The parameter is called `visit_func` because it suggests that as we "walk" around the directory, we "visit" each file.


def walk(dirname, visit_func):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        if os.path.isfile(path):
            visit_func(path)
        else:
            walk(path, visit_func)
```

```python
# Here's an example where we pass `print` as an argument, so when `walk` calls `visit_func`, it prints the paths of the files it discovers.

walk('photos', print)
```

## 13.9. Glossary
ephemeral: An ephemeral program typically runs for a short time and, when it ends, its data are lost.

persistent: A persistent program runs indefinitely and keeps at least some of its data in permanent storage.

directory: A collection of files and other directories.

current working directory: The default directory used by a program unless another directory is specified.

path: A string that specifies a sequence of directories, often leading to a file.

relative path: A path that starts from the current working directory, or some other specified directory.

absolute path: A path that does not depend on the current directory.

f-string: A string that has the letter f before the opening quotation mark, and contains one or more expressions in curly braces.

configuration data: Data, often stored in a file, that specifies what a program should do and how.

serialization: Converting an object to a string.

deserialization: Converting a string to an object.

database: A file whose contents are organized to perform certain operations efficiently.

key-value stores: A database whose contents are organized like a dictionary with keys that correspond to values.

binary mode: A way of opening a file so the contents are interpreted as sequence of bytes rather than a sequence of characters.

hash function: A function that takes and object and computes an integer, which is sometimes called a digest.

digest: The result of a hash function, especially when it is used to check whether two objects are the same.

### Answers for these Think Python questions
#### 1. What are the differences between ephemeral and persistent programs?
- **Ephemeral (transient) programs**: Run only in memory. All data is lost when the program exits or the computer shuts down. They do not save state to disk.
- **Persistent programs**: Store their data on storage (files, databases). Data survives after the program closes or the machine restarts.
> Example: A calculator script is ephemeral; a note-taking app that saves your notes is persistent.

#### 2. What are some examples of persistent programs?
Text editors, database servers, photo managers, web browsers (save bookmarks), password managers, Python programs using `shelve` / JSON files.

#### 3. What’s the difference between a relative path and an absolute path?
- **Absolute path**: Starts from the root of the filesystem, full complete location.
  Example: `/home/dinsdale/photos/photo.jpg` (macOS/Linux)
- **Relative path**: Relative to your current working directory; does not start at root.
  Example: `photos/photo.jpg`

#### 4. Why does the yaml module have functions called load and safe_load?
- `yaml.load()`: Can construct arbitrary Python objects. It is unsafe — a malicious YAML file can run arbitrary code.
- `yaml.safe_load()`: Only loads basic data types (dict, list, str, number). It blocks dangerous object creation, so it is the recommended default.

#### 5. When I write a Python shelf, what are the files with suffixes dat and dir?
`shelve` is built on top of `dbm`. It usually creates 3 files:
- `.dir`: Directory file, stores index information (key lookup table)
- `.dat`: Data file, stores the actual values
- `.bak`: Optional backup file
You should not edit these files directly; only access them via the `shelve` module.

#### 6. Other than key-value stores, what other kinds of databases are there?
Relational databases (SQLite, PostgreSQL, MySQL), document databases (MongoDB), columnar databases, graph databases, time-series databases.

#### 7. When I read a file, what’s the difference between binary mode and text mode?
- **Text mode (`'r'`, `'w'`)**: Reads data as strings. Python automatically handles newline conversion and text encoding (like UTF-8). Good for `.txt`, `.py`, `.yaml`.
- **Binary mode (`'rb'`, `'wb'`)**: Reads raw bytes. No encoding / newline translation. Required for images, PDFs, MD5 hash calculation.

#### 8. What are the differences between a bytes object and a string?
- `str`: Text, made of Unicode characters, for human-readable text. Example: `"abc"`
- `bytes`: Raw 8-bit binary data, sequence of numbers 0–255. Example: `b"abc"`
You convert between them with `.encode()` (str → bytes) and `.decode()` (bytes → str).

#### 9. What is a hash function?
A function that takes arbitrary-size input and returns a fixed-length string of bytes (hash / digest).
Properties:
- Same input always produces the same hash
- Small change in input → completely different hash
- It is one-way: you cannot recover the original input easily from the hash.

#### 10. What is an MD5 digest?
MD5 is a widely used hash function. An MD5 digest is its 128-bit hash value, usually written as a 32-character hexadecimal string.
Common use: check whether two files have identical content (your duplicate-photo exercise).
> Note: MD5 is cryptographically broken, but still fine for simple duplicate checking.

```python
# Original walk function from ThinkPython (for reference)
def walk(dirname):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        if os.path.isfile(path):
            print(path)
        elif os.path.isdir(path):
            walk(path)


def md5_digest(filename):
    data = open(filename, 'rb').read()
    md5_hash = hashlib.md5()
    md5_hash.update(data)
    digest = md5_hash.hexdigest()
    return digest

def same_contents(path1, path2):
    data1 = open(path1, 'rb').read()
    data2 = open(path2, 'rb').read()
    return data1 == data2

def is_image(path, extensions):
    """args: extensions is a list with photo's extension - config['extensions']
    returns: True if it is an image"""
    return os.path.splitext(path)[1][1:] in extensions


def md5_digest(filename):
    data = open(filename, 'rb').read()
    md5_hash = hashlib.md5()
    md5_hash.update(data)
    digest = md5_hash.hexdigest()  # this is a string
    return digest

def add_path(path, shelf):
    """
    Uses md5_digest to compute a digest for the path contents, then
    updates shelf to map a digest to a list of path
    """
    digest = md5_digest(path)  # digest is a string, uses function md5_digest
    if digest in shelf:
        path_list = shelf[digest]  # Be careful here, need to read and update old data, and write new data to shelf
        path_list.append(path)
        shelf[digest] = path_list
    else:
        shelf[digest] = [path]


def walk_images(dirname, shelf, extensions):
    """A version of walk, uses is_image to check if it is an image,
    if so, uses add_path to update a shelf"""
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        # Must check if it is a FILE first!
        if os.path.isfile(path) and is_image(path, extensions):
            add_path(path, shelf)

        elif os.path.isdir(path):
            print('\n')
            walk_images(path, shelf, extensions)


db = shelve.open('photos/digests', 'n')
walk_images('photos', db, config['extensions'])

# Important: close shelf to save data!
db.close()
```
